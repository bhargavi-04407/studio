```python
sales_occupation = df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
plt.figure(figsize=(20, 10))
sns.barplot(x='Occupation', y='Amount', hue='Occupation', data=sales_occupation, legend=False)
plt.title('Total Amount by Occupation')
plt.xlabel('Occupation')
plt.ylabel('Total Amount')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
sales_occupation = df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
plt.figure(figsize=(20, 10))
sns.barplot(x='Occupation', y='Amount', hue='Amount', data=sales_occupation, legend=False)
plt.title('Total Amount by Occupation')
plt.xlabel('Occupation')
plt.ylabel('Total Amount')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
sales_product_category = df.groupby(['Product_Category'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
plt.figure(figsize=(20, 10))
sns.barplot(x='Product_Category', y='Amount', data=sales_product_category)
plt.title('Total Amount by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Amount')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
total_categories = df['Product_Category'].nunique()
total_amount = df['Amount'].sum()

print(f"Total Number of Product Categories: {total_categories}")
print(f"Total Sales Amount: {total_amount}")
```

```python
sales_product_category_counts = df.groupby(['Product_Category']).size().reset_index(name='CountOfTransactions')
sales_product_category_counts_sorted = sales_product_category_counts.sort_values(by='CountOfTransactions', ascending=False)

plt.figure(figsize=(20, 10))
sns.barplot(x='Product_Category', y='CountOfTransactions', data=sales_product_category_counts_sorted)
plt.title('Count of Transactions by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Count of Transactions')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
sales_marital_gender = df.groupby(['Product_Category', 'Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
plt.figure(figsize=(20, 10))
sns.barplot(x='Product_Category', y='Amount', hue='Gender', data=sales_marital_gender)
plt.title('Total Amount by Product Category and Gender')
plt.xlabel('Product Category')
plt.ylabel('Total Amount')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
sales_product_id_orders = df.groupby(['Product_ID'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)

plt.figure(figsize=(15, 7))
sns.barplot(x='Product_ID', y='Orders', data=sales_product_id_orders)
plt.title('Top 10 Products by Total Orders')
plt.xlabel('Product ID')
plt.ylabel('Total Orders')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
sales_product_id_amount = df.groupby(['Product_ID'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)

plt.figure(figsize=(15, 7))
sns.barplot(x='Product_ID', y='Amount', data=sales_product_id_amount)
plt.title('Top 10 Products by Total Sales Amount')
plt.xlabel('Product ID')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
# Calculate sales by State
sales_by_state = df.groupby('State')['Amount'].sum().sort_values(ascending=False).reset_index()

# Calculate sales by Occupation
sales_by_occupation = df.groupby('Occupation')['Amount'].sum().sort_values(ascending=False).reset_index()

# Create subplots
plt.figure(figsize=(20, 8))

# Plot for Sales by State
plt.subplot(1, 2, 1)  # 1 row, 2 columns, 1st plot
sns.barplot(x='State', y='Amount', data=sales_by_state.head(10)) # Limiting to top 10 states for readability
plt.title('Top 10 States by Total Sales Amount')
plt.xlabel('State')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=45, ha='right')

# Plot for Sales by Occupation
plt.subplot(1, 2, 2)  # 1 row, 2 columns, 2nd plot
sns.barplot(x='Occupation', y='Amount', data=sales_by_occupation.head(10)) # Limiting to top 10 occupations for readability
plt.title('Top 10 Occupations by Total Sales Amount')
plt.xlabel('Occupation')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```
